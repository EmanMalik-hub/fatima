```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,classification_report,precision_score,recall_score
data={
    "links":[0,5,7,9],
    "length":[20,30,40,50],
     "spam":[0,0,1,1]
}
df=pd.DataFrame(data)
print(df)
X=df[["links","length"]]
Y=df["spam"]
model=LogisticRegression()
model.fit(X,Y)




print(model.coef_)
print(model.intercept_)
#spam emailchec k karna
spam_email=pd.DataFrame([[11,60]],columns=["links","length"])
print(model.predict(spam_email))
print("spam email")
print(model.predict_proba(spam_email))
# checking the normal email
normal_email=pd.DataFrame([[1,15]],columns=["links","length"])
print("normal emial")
print(model.predict(normal_email))
print(model.predict_proba(normal_email))
#middl email
middle_mail=pd.DataFrame([[6,35]],columns=["links","length"])
print(model.predict(middle_mail))
print(model.predict_proba(middle_mail))
Y_predict=model.predict(X)
print(confusion_matrix(Y,Y_predict))
print(precision_score(Y,Y_predict))
print(recall_score(Y,Y_predict))
```
